```python
import torch
from transformers import BertTokenizerFast, AutoTokenizer
```

```python
sentence = [
    'I love my dog',
    'f hate my cat'
]

```

```python
def tokenize(text):
    """
    Token the provided text by converting into lowercase
    splitting it whitespace
    """
    #Lowercase the txt and split by whotespace
    tokens = text.lower().split()

    return tokens
```

```python
tokens = tokenize(sentence[1])
print(tokens)
```

```python
def build_vocab(sentences):
  """
  CONSTRUCT A VOCAB MAPPING FROM A LIST
  ASSIGN UNQIE INT
  """

  vocabs ={}

  for sentence in sentences:
    tokens = tokenize(sentence)

    for token in tokens:
      if token not in vocabs:
        vocabs[token] = len(vocabs) + 1
  return vocabs
```

```python
vocab = build_vocab(sentence)

print("Vocabulary Index:", vocab, "\n")
```

```python
sentences = [
    'I love my dog !@',
    'I love my cat'
]
```

```python
def print_unique_token_id_map(tokens_list,input_ids_tensor):
    """
    Prints unique token to ID mappings from a list of tokenized sentences,
    sorted by token ID.
    """
    unique_token_to_id_map = {}
    for i in range(len(tokens_list)):
        sentence_tokens = tokens_list[i]

        sentence_ids = input_ids_tensor[i]
        for j in range(len(sentence_tokens)):
            token_str = sentence_tokens[j]

            token_id =sentence_ids[j].item()
            unique_token_to_id_map[token_str] = token_id


    sorted_token_mapping = sorted(unique_token_to_id_map.items(),key=lambda item:item[i])
    print("\n--- Unique Token to ID Mappings (for these sentences) ---")
    if not sorted_token_mapping:
        print("No tokens to display")
        return

    for token, token_id in sorted_token_mapping:
      print(f"{token}\t-->\t{token_id}")

    print()
```

```python
tokenizer = BertTokenizerFast.from_pretrained("google-bert/bert-base-uncased")
#encode n tokenize the sentence
encoded_inputs = tokenizer(sentences, padding=True , trucation=True,return_tensors='pt')
tokens = [tokenizer.convert_ids_to_tokens(ids)
         for ids in encoded_inputs["input_ids"]]
word_index = tokenizer.get_vocab()

print("tokens:",tokens)
print("\n token ids:", encoded_inputs['input_ids'])
print_unique_token_id_map(tokens, encoded_inputs['input_ids'])
```
