```python
def towerhanoi(n,source,target,auxiliary):
  if n == 1:
    print("move disk 1 from source",source,"to target",target)
  return
  towerhanoi(n-1,source,auxiliary,target)
  print("move disk",n,"from source",source,"to target",target)
  towerhanoi(n-1,auxiliary,target,source)
```
